### Bài 1: Tìm số lớn nhất của 1 list. KHONG DUNG HAM SAN CO CUA LIST

```python
def max_of_list(lst):
    if not lst:
        return None
    max_val = lst[0]
    for x in lst[1:]:
        if x > max_val:
            max_val = x
    return max_val

assert max_of_list([]) == None
assert max_of_list([23, 67, 900, -10]) == 900
assert max_of_list([575, 939, 697, 138, 183, 386, 147, 695, 622, 289]) == 939

print("Pass all Tests!")
```

### Bài 2: Tìm số nhỏ nhất của một danh sách các số nguyên. KHONG DUNG HAM SAN CO CUA LIST

```python
def min_of_list(lst):
    if not lst:
        return None
    min_val = lst[0]
    for x in lst[1:]:
        if x < min_val:
            min_val = x
    return min_val

assert min_of_list([]) == None
assert min_of_list([23, 67, 900, -10]) == -10
assert min_of_list([575, 939, 697, 138, 183, 386, 147, 695, 622, 289]) == 138

print("Pass all Tests!")
```

### Bài 2b: Tìm số nhỏ nhất của 1 list và vị trí của nó, nếu có nhiều số nhỏ nhất bằng nhau thì lấy chỉ số đầu tiên (nhỏ nhất)

```python
def min_and_pos(lst):
    if not lst:
        return None
    min_val = lst[0]
    min_idx = 0
    for i in range(1, len(lst)):
        if lst[i] < min_val:
            min_val = lst[i]
            min_idx = i
    return (min_val, min_idx)

assert min_and_pos([10, 0, 0, 12, 34, 5, 6, 7, 8, 89, 0, 1]) == (0, 1)
assert min_and_pos([]) == None

print("Pass all Tests!!!")
```

### Bài 3: Tìm giá trị trung bình của các số nguyên dương trong một list.
Nếu không có số dương nào thì trả về "invalid"

```python
def Average_positive_numbers_of_list(lst):
    positives = [x for x in lst if x > 0]
    if not positives:
        return "invalid"
    return round(sum(positives) / len(positives), 2)

# Alias để test
Average_lst = Average_positive_numbers_of_list

assert Average_lst([1, 2, 6, 1, -10, -10]) == 2.50
assert Average_lst([-2, -1, -20, -343]) == "invalid"
assert Average_lst([]) == "invalid"

print("Accepted!")
```

### Bài 4: Kiểm tra một list số nguyên có phải là đối xứng hay không

```python
def kt_doi_xung(lst):
    if not lst:
        return None
    return lst == lst[::-1]

assert kt_doi_xung([1, 2, 3, 5, 3, 2, 1]) == True
assert kt_doi_xung([1, 2, 3, 5, 3, 4, 1]) != True
assert kt_doi_xung([]) == None

print("Accepted!!")
```

### Bài 5: Đưa tất cả các số bằng 0 trong list về đầu của list. Các số khác không thay đổi thứ tự trước sau

```python
def move_zero_to_first(lst):
    zeros = [x for x in lst if x == 0]
    non_zeros = [x for x in lst if x != 0]
    return zeros + non_zeros

assert move_zero_to_first([3, 4, 0, 0, 23, 0, 0, 34, 67, 89]) == [0, 0, 0, 0, 3, 4, 23, 34, 67, 89]
assert move_zero_to_first([0, 0, 0, 0]) == [0, 0, 0, 0]
assert move_zero_to_first([3, 4, 0, 0, 1, 2, 3]) == [0, 0, 3, 4, 1, 2, 3]

print("accepted!!!")
```

### Bài 6: Xóa tất cả các số giống nhau trong list, giữ lại đúng 1 lần, Các số khác giữ nguyên vị trí trước

```python
def unique_array(lst):
    if not lst:
        return None
    seen = []
    for x in lst:
        if x not in seen:
            seen.append(x)
    return seen

assert unique_array([1, 2, 3, 4, 1, 2, 3, 4, 1, 1, 4, 3]) == [1, 2, 3, 4]
assert unique_array([1, 1, 1, 1]) == [1]
assert unique_array([]) == None

print("Accepted!!!")
```

### Bài 7: Kiểm tra một list số nguyên có phải là dãy tăng, dãy giảm hay vô danh

```python
def check_order_of_list(lst):
    if len(lst) <= 1:
        return "Tang"
    tang = all(lst[i] < lst[i+1] for i in range(len(lst)-1))
    giam = all(lst[i] > lst[i+1] for i in range(len(lst)-1))
    if tang:
        return "Tang"
    elif giam:
        return "Giam"
    else:
        return "Vo danh"

assert check_order_of_list([1, 3, 5, 6, 7, 8, 9]) == "Tang"
assert check_order_of_list([9, 8, 7, 6, 5, 4, 0]) == "Giam"
assert check_order_of_list([1, 2, 3, 7, 1]) == "Vo danh"
assert check_order_of_list([7, 6, 5, 4, 3, 3]) == "Vo danh"

print("Pass all tests!!!")
```

### Bài 8: Kiểm tra một danh sách số nguyên có phải là ziczac không

```python
def is_ziczac(lst):
    if len(lst) <= 1:
        return True
    for i in range(1, len(lst) - 1):
        if not ((lst[i] > lst[i-1] and lst[i] > lst[i+1]) or
                (lst[i] < lst[i-1] and lst[i] < lst[i+1])):
            return False
    return True

assert is_ziczac([1, 3, 2, 5, 0, 7, 6, 8, 0]) == True
assert is_ziczac([34, 2, 56, 7, 80, 12, 24, 14]) != False
assert is_ziczac([2, 2, 34, 4, 56, 7, 89]) == False

print("All accepted!")
```

### Bài 9: Tìm và in ra tất cả các vị trí của số nguyên k trong một danh sách số nguyên cho trước.

```python
def find_pos_of_k(lst, k):
    positions = [i for i, x in enumerate(lst) if x == k]
    return positions if positions else None

assert find_pos_of_k([1,3,4,4,3,3,4,3,6,1], 3) == [1, 4, 5, 7]
assert find_pos_of_k([1,3,4,4,3,3,4,3,6,1], 10) == None
assert find_pos_of_k([1, 1, 2, 3], 3) == [3]

print("Tests accepted!")
```

### Bài 10: Nhập vào tên một người, in ra kí tự đầu tiên và kí tự cuối cùng của tên đó.

```python
def theFirst_Last_character(s):
    if not s:
        return None
    return [s[0], s[-1]]

assert theFirst_Last_character("Nguyen Ai Quoc") == ["N", "c"]
assert theFirst_Last_character("Ton Duc Thang") == ["T", "g"]
assert theFirst_Last_character("") == None

print("passed all tests!")
```

### Bài 11: Liệt kê số lần xuất hiện của các số có mặt trong danh sách. Số nào xuất hiện trước thì liệt kê trước

```python
def list_number_in_lst(lst):
    if not lst:
        return None
    result = {}
    for x in lst:
        if x not in result:
            result[x] = 0
        result[x] += 1
    return result

assert list_number_in_lst([1, 2, 1, 1, 2, 1, 1]) == {1: 5, 2: 2}
assert list_number_in_lst([1, 2, 1, 1, 2, 1, 1, 3, 3, 4, 2, 1, 2, 3]) == {1: 6, 2: 4, 3: 3, 4: 1}
assert list_number_in_lst([]) == None
assert list_number_in_lst([4, 3, 1, 2, 1, 1, 2, 1, 1]) == {4: 1, 3: 1, 1: 5, 2: 2}

print("Pass all tests!")
```

### Bài 11B: Liệt kê số lần xuất hiện - Số nào nhỏ thì liệt kê trước (sắp xếp theo số lần xuất hiện tăng dần)

```python
def list_number_in_lst_order(lst):
    if not lst:
        return None
    # Đếm số lần xuất hiện, giữ thứ tự xuất hiện đầu tiên
    count = {}
    for x in lst:
        if x not in count:
            count[x] = 0
        count[x] += 1
    # Sắp xếp theo số lần xuất hiện tăng dần
    return dict(sorted(count.items(), key=lambda item: item[1]))

assert list_number_in_lst_order([2, 3, 3, 1, 1, 2, 1]) == {2: 2, 3: 2, 1: 3}
assert list_number_in_lst_order([2, 3, 8, 8, 8, 3, 1, 1, 2, 1, 4, 3, 4, 5]) == {2: 2, 3: 3, 8: 3, 1: 3, 4: 2, 5: 1}

print("Dung roi")
```

### Bài 12: Liệt kê số lần xuất hiện của các kí tự có mặt trong chuỗi theo thứ tự tăng dần.

```python
def count_character_of_string(s):
    if not s:
        return None
    count = {}
    for c in s:
        if c not in count:
            count[c] = 0
        count[c] += 1
    # Sắp xếp theo ký tự tăng dần (alphabetical)
    return dict(sorted(count.items()))

assert count_character_of_string("abcabccc") == {'a': 2, 'b': 2, 'c': 4}
assert count_character_of_string("fbdfgfhsgfjsghghg") == {'b': 1, 'd': 1, 'f': 4, 'g': 5, 'h': 3, 'j': 1, 's': 2}

print("Passed!")
```
